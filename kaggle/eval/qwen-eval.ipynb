```python
import os
import time

os.environ['HF_HUB_ENABLE_HF_TRANSFER'] = '1'
from huggingface_hub import hf_hub_download

t0 = time.time()
GGUF_PATH = hf_hub_download(repo_id=REPO, filename=GGUF_FILE)
print(f'downloaded in {(time.time() - t0) / 60:.1f} min:', GGUF_PATH)
```

```python
import json
import subprocess
import time
import urllib.request

log = open('/tmp/llama-server.log', 'w')
cmd = [
    'python', '-m', 'llama_cpp.server',
    '--model', GGUF_PATH,
    '--n_gpu_layers', str(N_GPU_LAYERS),
    '--n_ctx', str(N_CTX),
    '--host', '127.0.0.1',
    '--port', str(PORT),
]
print('launch:', ' '.join(cmd))
proc = subprocess.Popen(cmd, stdout=log, stderr=subprocess.STDOUT)
for i in range(60):
    if proc.poll() is not None:
        print('--- server died on startup; log tail ---')
        print(''.join(open('/tmp/llama-server.log').readlines()[-40:]))
        raise RuntimeError(f"server exited with code {proc.returncode}")
    time.sleep(10)
    try:
        with urllib.request.urlopen(f'http://127.0.0.1:{PORT}/v1/models', timeout=5) as r:
            print('READY, status', r.status)
            break
    except Exception:
        if i % 6 == 0:
            print(f'waiting for server... ({i * 10}s)')
else:
    print('--- server log tail ---')
    print(''.join(open('/tmp/llama-server.log').readlines()[-30:]))
    raise RuntimeError('server did not come up in 10 min')
```

```python
import json
import urllib.request


def chat(content, max_tokens=50):
    body = json.dumps({
        'messages': [{'role': 'user', 'content': content}],
        'temperature': 0,
        'max_tokens': max_tokens,
    }).encode()
    req = urllib.request.Request(
        f'http://127.0.0.1:{PORT}/v1/chat/completions',
        data=body,
        headers={'Content-Type': 'application/json'},
    )
    with urllib.request.urlopen(req, timeout=300) as r:
        return json.load(r)


d = chat('Reply with exactly: ok')
text = d['choices'][0]['message']['content']
print('reply:', repr(text[:120]))
assert text and text.strip(), 'empty reply — model misbehaving; check /tmp/llama-server.log'
assert d['choices'][0]['finish_reason'] == 'stop', d['choices'][0]['finish_reason']
print('SMOKE OK')
```

```python
import json

D = json.load(open('/kaggle/input/qwen-eval-irs/irs.json'))
P = json.load(open('/kaggle/input/qwen-eval-irs/prompts.json'))
assert len(D) == 45, len(D)
assert set(P) == {'v0', 'strict'}
assert all('{draft}' in P[a] and 'TESTDRAFT' not in P[a] for a in P)
print('dataset OK:', len(D), 'IRs + 2 prompt templates')
```

```python
import json
import time
import urllib.request

OUT = '/kaggle/working/replies.jsonl'
ERRORS = 0
N = 0
out = open(OUT, 'w')
t0 = time.time()
for i, item in enumerate(D):
    for arm in ('v0', 'strict'):
        prompt = P[arm].replace('{draft}', item['ir'])
        assert '{draft}' not in prompt and 'TESTDRAFT' not in prompt
        body = json.dumps({
            'messages': [{'role': 'user', 'content': prompt}],
            'temperature': 0,
            'max_tokens': 512,
        }).encode()
        row = {'key': item['key'], 'arm': arm, 'reply': '',
               'finish': '', 'error': None}
        ok = False
        for attempt in range(3):
            try:
                req = urllib.request.Request(
                    f'http://127.0.0.1:{PORT}/v1/chat/completions',
                    data=body, headers={
                        'Content-Type': 'application/json'},
                )
                with urllib.request.urlopen(req, timeout=600) as r:
                    d = json.load(r)
                row['reply'] = d['choices'][0]['message']['content'] or ''
                row['finish'] = d['choices'][0]['finish_reason']
                ok = True
                break
            except Exception as e:
                row['error'] = f'{type(e).__name__}: {e}'[:200]
                time.sleep(10)
        if not ok:
            ERRORS += 1
        out.write(json.dumps(row) + '\n')
        out.flush()
        N += 1
        if N % 10 == 0:
            print(f'{N}/90 rows, {ERRORS} errors', flush=True)
        assert ERRORS <= 10, f'error budget exceeded ({ERRORS})'
print(f'done: {N} rows, {ERRORS} errors in {(time.time() - t0) / 60:.1f} min')
```
