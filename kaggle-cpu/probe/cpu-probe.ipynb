# CPU spec probe (instant, free)
Answers: RAM, cores, disk, python version. Run 027.

```python
import subprocess


def sh(cmd):
    r = subprocess.run(cmd, shell=True, capture_output=True, text=True)
    print(f'$ {cmd}  (exit {r.returncode})')
    print((r.stdout + r.stderr)[:1200])
    print('---')


sh('free -g; nproc; df -h /tmp /kaggle/working 2>/dev/null')
sh('python3 --version; pip --version 2>&1 | head -1')
sh('cat /proc/cpuinfo | grep -m1 "model name"')
```
